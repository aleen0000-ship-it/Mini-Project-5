```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.feature_selection import VarianceThreshold

df = pd.read_csv("/kaggle/input/datasets/aleenf/projectfhehofuw/day25_project.csv")

target_col = df.columns[-1]

X = df.drop(target_col, axis=1)
y = df[target_col]

num_cols = X.select_dtypes(include=np.number).columns

for col in num_cols:
    X[col + "_log"] = np.log1p(X[col])
    X[col + "_sqrt"] = np.sqrt(X[col])

poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X[num_cols])
X_poly = pd.DataFrame(X_poly)

X = pd.concat([X.reset_index(drop=True), X_poly.reset_index(drop=True)], axis=1)

X.columns = X.columns.astype(str)

selector = VarianceThreshold(threshold=0.01)
X_selected = selector.fit_transform(X)
X_selected = pd.DataFrame(X_selected)

df_final = pd.concat([X_selected, y.reset_index(drop=True)], axis=1)

df_final.to_csv("/kaggle/working/day25_engineered.csv", index=False)
```
